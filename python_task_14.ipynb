```python
text_columns = ["Country", "Region"]
for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

print("Text columns cleaned successfully.")
```

```python
print("Negative life expectancy values:", (df["Life_expectancy"] < 0).sum())
print("Negative GDP per capita values:", (df["GDP_per_capita"] < 0).sum())
print("Invalid years:", (~df["Year"].between(1900, 2100)).sum())
```

```python
# Basic dataset overview
record_count = len(df)
country_count = df["Country"].nunique()
region_count = df["Region"].nunique()
print("Number of country-year records:", record_count)
print("Number of countries:", country_count)
print("Number of regions:", region_count)
```

```python
average_life_expectancy = df["Life_expectancy"].mean()
print("Average life expectancy:", round(average_life_expectancy, 2), "years")
```

```python
highest_life_expectancy = df["Life_expectancy"].max()
print("Highest recorded life expectancy:", highest_life_expectancy, "years")
```

```python
lowest_life_expectancy = df["Life_expectancy"].min()
print("Lowest recorded life expectancy:", lowest_life_expectancy, "years")
```

```python
year_count = df["Year"].nunique()
print("Number of years covered:", year_count)
```

```python
average_schooling = df["Schooling"].mean()
print("Average schooling measure:", round(average_schooling, 2))
```

```python
# Grouped summary 1: by region
region_summary = df.groupby("Region").agg(
    Records=("Life_expectancy", "size"),
    Countries=("Country", "nunique"),
    Average_Life_Expectancy=("Life_expectancy", "mean"),
    Average_GDP_per_Capita=("GDP_per_capita", "mean"),
    Average_Schooling=("Schooling", "mean")
).sort_values("Average_Life_Expectancy", ascending=False)

region_summary
```

```python
# Grouped summary 2: by year
year_summary = df.groupby("Year").agg(
    Records=("Life_expectancy", "size"),
    Average_Life_Expectancy=("Life_expectancy", "mean"),
    Average_GDP_per_Capita=("GDP_per_capita", "mean"),
    Average_Infant_Deaths=("Infant_deaths", "mean"),
    Average_Under_Five_Deaths=("Under_five_deaths", "mean")
).sort_index()

year_summary
```

```python
# Grouped summary 3: developed vs developing
df["Economy_Status"] = df["Economy_status_Developed"].map({1: "Developed", 0: "Developing"})
economy_summary = df.groupby("Economy_Status").agg(
    Records=("Life_expectancy", "size"),
    Countries=("Country", "nunique"),
    Average_Life_Expectancy=("Life_expectancy", "mean"),
    Median_Life_Expectancy=("Life_expectancy", "median"),
    Average_GDP_per_Capita=("GDP_per_capita", "mean"),
    Average_Schooling=("Schooling", "mean")
).sort_values("Average_Life_Expectancy", ascending=False)

economy_summary
```

```python
# Chart 1: average life expectancy by year
plt.figure(figsize=(12, 6))
sns.lineplot(x=year_summary.index, y=year_summary["Average_Life_Expectancy"], marker="o", color="green")
plt.title("Average Life Expectancy by Year")
plt.xlabel("Year")
plt.ylabel("Life expectancy (years)")
plt.show()
```

```python
# Chart 2: average life expectancy by region
region_summary["Average_Life_Expectancy"].sort_values().plot(kind="barh", color="teal")
plt.title("Average Life Expectancy by Region")
plt.xlabel("Life expectancy (years)")
plt.ylabel("Region")
plt.show()
```

```python
# Chart 3: average life expectancy by economy status
economy_summary["Average_Life_Expectancy"].plot(kind="bar", color="purple")
plt.title("Average Life Expectancy by Economy Status")
plt.xlabel("Economy status")
plt.ylabel("Life expectancy (years)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Chart 4: GDP per capita and life expectancy
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="GDP_per_capita", y="Life_expectancy", alpha=0.5)
plt.title("GDP per Capita and Life Expectancy")
plt.xlabel("GDP per capita (source units)")
plt.ylabel("Life expectancy (years)")
plt.show()
```

```python
# Chart 5: schooling and life expectancy
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="Schooling", y="Life_expectancy", alpha=0.5, color="darkorange")
plt.title("Schooling and Life Expectancy")
plt.xlabel("Schooling (source measure)")
plt.ylabel("Life expectancy (years)")
plt.show()
```

```python
# Chart 6: infant deaths and life expectancy
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="Infant_deaths", y="Life_expectancy", alpha=0.5, color="firebrick")
plt.title("Infant Deaths and Life Expectancy")
plt.xlabel("Infant deaths (source measure)")
plt.ylabel("Life expectancy (years)")
plt.show()
```

```python
# Chart 7: average immunization measures by region
vaccination_summary = df.groupby("Region")[["Hepatitis_B", "Polio", "Diphtheria"]].mean()
vaccination_summary.plot(kind="bar", figsize=(6, 1))
plt.title("Average Immunization Measures by Region")
plt.xlabel("Region")
plt.ylabel("Measure (source scale)")
plt.xticks(rotation=35)
plt.show()
```

```python
# Chart 8: under-five deaths and life expectancy
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="Under_five_deaths", y="Life_expectancy", alpha=0.5, color="brown")
plt.title("Under-Five Deaths and Life Expectancy")
plt.xlabel("Under-five deaths (source measure)")
plt.ylabel("Life expectancy (years)")
plt.show()
```

```python
# Dashboard KPIs
print("========== LIFE EXPECTANCY DASHBOARD KPIs ==========")
print("Country-year records:", record_count)
print("Countries:", country_count)
print("Regions:", region_count)
print("Average life expectancy:", round(average_life_expectancy, 2), "years")
print("Highest life expectancy:", highest_life_expectancy, "years")
print("Lowest life expectancy:", lowest_life_expectancy, "years")
print("Average schooling measure:", round(average_schooling, 2))
```

```python
top_region = region_summary["Average_Life_Expectancy"].idxmax()
top_region_value = region_summary["Average_Life_Expectancy"].max()
print("Highest regional average:", top_region)
print("Average life expectancy:", round(top_region_value, 2), "years")
```

```python
top_country = df.groupby("Country")["Life_expectancy"].mean().idxmax()
top_country_value = df.groupby("Country")["Life_expectancy"].mean().max()
print("Country with highest average in this dataset:", top_country)
print("Average life expectancy:", round(top_country_value, 2), "years")
```

```python
year_with_highest_mean = year_summary["Average_Life_Expectancy"].idxmax()
highest_year_mean = year_summary["Average_Life_Expectancy"].max()
print("Year with highest dataset-wide average:", year_with_highest_mean)
print("Average life expectancy:", round(highest_year_mean, 2), "years")
```

```python
correlations = df[["GDP_per_capita", "Schooling", "Infant_deaths", "Life_expectancy"]].corr()["Life_expectancy"]
print("Correlation with life expectancy:")
print(correlations.drop("Life_expectancy").round(3))
print("Note: correlation describes association, not causation.")
```
